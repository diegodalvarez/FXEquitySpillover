```python
group_mapper = {
    'em_carry'       : "EM Carry", 
    'g10_carry'      : "G10 Carry", 
    'illiquid_future': "Illiquid Fut", 
    'liquid_future'  : "Liquid Fut", 
    'full'           : "Full Portfolio"}

tmp_renamer = {
    "perf_rtn": "Vol. Adj",
    "px_rtn"  : "Raw Return",
    "lagged"  : "Lagged",
    "perfect" : "Contemporaneous"}

title_renamer = {
    "in_sample"         : "In-Sample",
    "out_sample"        : "Out-Sample",
    "ReturnSpreadSignal": "ETF Returns Differential",
    "ReturnsSignal"     : "ETF Returns"}

factor_renamer = {
    'BEST_CALCULATED_FCF'  : 'Est FCF',
    'BEST_CUR_EV_TO_EBITDA': 'Est EV/EBITDA',
    'BEST_DIV_YLD'         : 'Est Div Yld',
    'BEST_EBIT'            : 'Est EBIT',
    'BEST_EPS'             : 'Est EPS',
    'BEST_GROSS_MARGIN'    : 'Est Gross Margin',
    'BEST_PE_RATIO'        : 'Est P/E',
    'BEST_PX_SALES_RATIO'  : 'Est P/Sales',
    'BEST_TARGET_PRICE'    : 'Est Price',
    'CURR_ENTP_VAL'        : 'Curr EV',
    'FREE_CASH_FLOW_YIELD' : 'FCF',
    'FUND_FLOW'            : 'Fund Flow',
    'NET_AGGTE_DVD_YLD'    : 'Div Yld',
    'PE_RATIO'             : 'P/E',
    'PX_TO_SALES_RATIO'    : 'P/Sales',
    'PX_VOLUME'            :'Volume'}

lag_renamer = {
    "no_lag"     : "No Lag",
    "with_lag"   : "With Lag",
    "expanding"  : "Expanding",
    "full_sample": "Full-Sample",
    "train_test" : "Train/Test",
    "lag_vol_rtn": "Lagged",
    "vol_rtn"    : "Perfect"}

param_renamer = {
    "beta"    : "Coeff",
    "pvalue"  : "$p$-value",
    "r2"      : "$R^2$",
    "tvalue"  : "$t$-value",
    "stat_sig": "Stat. Sig.",	
    "exp"     : "Expected",	
    "ratio"   : "FDR Sig"}

RENAMER = {**group_mapper, **title_renamer, **tmp_renamer, **factor_renamer, **param_renamer, **lag_renamer}
FLOAT_FORMAT = lambda x: f"{x:.2f}"
```

# Examining Factor Data

```python
path   = r"C:\Users\Diego\Documents\GitHub\FXEquitySpillover\data\PrepData\FactorData.parquet"
df_tmp = (pd
    .read_parquet(path = path, engine = "pyarrow")
    .loc[lambda x: x.field == "BEST_PE_RATIO"])

fig, axes = plt.subplots(ncols = 2, figsize = (20,6))

(df_tmp
    .loc[lambda x: x.calc == "adjusted"]
    .pivot(index = "date", columns = "etf_ticker", values = "value")
    .cumsum()
    .plot(
        ax     = axes[0],
        legend = False, 
        alpha  = 0.7,
        title  = "Adjusted Value"))

(df_tmp
    .loc[lambda x: x.calc == "raw"]
    .pivot(index = "date", columns = "etf_ticker", values = "value")
    .plot(
        ax     = axes[1],
        legend = False,
        alpha  = 0.7,
        title  = "Raw Value"))

for ax in axes.flatten(): ax.set_ylabel("Est. P/E Ratio")
fig.suptitle("Comparing Adjusted and Raw Estimated P/E Ratio")
#fig.savefig(os.path.join(pic_path, "FundamentalFactorDataAdj.jpeg"), dpi = 300)
plt.show()
```

# Analyze Factor Models

```python
path   = r"C:\Users\Diego\Documents\GitHub\FXEquitySpillover\data\FactorAnalysis\ETFFactors.parquet"
df_params = (pd
    .read_parquet(path = path, engine = "pyarrow")
    .loc[lambda x: x.vol_target == "perfect"])

df_prep = (df_params
    .loc[lambda x: x.calc == "adjusted"]
    .loc[lambda x: x.signal_lag == "no_lag"]
    .drop(columns = ["vol_target", "calc", "source", "signal_lag", "n"]))
```

```python
df_ratio = (df_prep
    .assign(stat_sig = lambda x: np.where((np.abs(x.tvalue) > 1.5) & (x.pvalue < 0.05), "stat_sig", "not_sig"))
    [["field", "stat_sig", "fx_ticker"]]
    .groupby(["field", "stat_sig"])
    .agg("count")
    .reset_index()
    .pivot(index = "field", columns = "stat_sig", values = "fx_ticker")
    .assign(
        exp   = lambda x: (x.not_sig + x.stat_sig) * 0.05,
        ratio = lambda x: x.stat_sig / (x.stat_sig + x.not_sig))
    .drop(columns = ["not_sig"])
    .reset_index())
```

```python
display(df_prep
    .melt(id_vars = ["fx_ticker", "etf_ticker", "field"])
    [["field", "variable", "value"]]
    .assign(value = lambda x: x.value.abs())
    .groupby(["field", "variable"])
    .agg("mean")
    .reset_index()
    .pivot(index = "field", columns = "variable", values = "value")
    .merge(right = df_ratio, how = "inner", on = ["field"])
    .replace(RENAMER)
    .rename(columns = param_renamer)
    .set_index("field"))
```

```python
df_compare = (df_params
    .loc[lambda x: x.signal_lag == "no_lag"]
    .drop(columns = ["signal_lag", "vol_target", "source","n"])
    .melt(id_vars = ["fx_ticker", "etf_ticker", "calc", "field"])
    [["calc", "field", "variable", "value"]]
    .assign(value = lambda x: np.abs(x.value)))
```

```python
display(df_compare
    .groupby(["calc", "field", "variable"])
    .agg("mean")
    .reset_index()
    .replace(RENAMER)
    .pivot(index = "field", columns = ["calc", "variable"], values = "value"))
```

```python
(df_params
    .loc[lambda x: x.calc == "adjusted"]
    .drop(columns = ["n", "vol_target", "source", "calc"])
    .melt(id_vars = ["fx_ticker", "etf_ticker", "field", "signal_lag"])
    [["field", "signal_lag", "variable", "value"]]
    .assign(value = lambda x: x.value.abs())
    .groupby(["field", "signal_lag", "variable"])
    .agg("mean")
    .reset_index()
    .replace(RENAMER)
    .rename(columns = {
        "signal_lag": ""})
    .pivot(index = ["field"], columns = ["", "variable"], values = "value"))
```

```python
path = r"C:\Users\Diego\Documents\GitHub\FXEquitySpillover\data\Backtests\FactorForecastedTrend.parquet"
df_group_port = (pd
    .read_parquet(path = path, engine = "pyarrow")
    [["date", "field", "optimization", "fx_ticker", "lag_vol_rtn", "vol_rtn"]]
    .melt(id_vars = ["date", "field", "optimization", "fx_ticker"])
    .drop(columns = ["fx_ticker"])
    .groupby(["date", "field", "optimization", "variable"])
    .agg("mean")
    .reset_index())
```

```python
fig, axes = plt.subplots(figsize = (8,5))

df_plot = (df_group_port
    .loc[lambda x: x.variable == "vol_rtn"]
    .loc[lambda x: x.optimization == "full_sample"]
    .rename(columns = {"field": ""})
    .pivot(index = "date", columns = "", values = "value")
    .assign(avg_val = lambda x: x.mean(axis = 1))
    .rename(columns = RENAMER)
    .apply(lambda x: np.cumprod(1 + x)))

(df_plot
    .drop(columns = ["avg_val"])
    .plot(
        logy   = True,
        ax     = axes, 
        legend = False,
        alpha  = 0.7))

(df_plot
    [["avg_val"]]
    .rename(columns = {"avg_val": "Avg."})
    .plot(
        logy   = True,
        ax     = axes, 
        color  = "black",
        ylabel = "Cuml. Ret. Log-Scaled",
        title  = "Cumulative Returns of Full-Sample In-Sample OLS Forecasted Values\n10% Annualized Volatility Targeting Hedged Contemporaneously"))

#fig.savefig(os.path.join(pic_path, "FactorForecastedBacktest.jpeg"), dpi = 300)
plt.show()
```

```python
display(df_group_port
    .loc[lambda x: x.variable == "vol_rtn"]
    .drop(columns = ["variable"])
    .pivot(index = ["date", "optimization"], columns = "field", values = "value")
    .assign(avg_val = lambda x: x.mean(axis = 1))
    .reset_index()
    .melt(id_vars = ["date", "optimization"])
    .dropna()
    .drop(columns = ["date"])
    .groupby(["optimization", "field"])
    .agg(lambda x: x.mean() / x.std() * np.sqrt(252))
    .reset_index()
    .replace(RENAMER)
    .replace({"avg_val": "Avg"})
    .pivot(index = "field", columns = "optimization", values = "value"))
```

```python
fig, axes = plt.subplots(figsize = (8,5))

(df_group_port
    .loc[lambda x: x.variable == "vol_rtn"]
    .drop(columns = ["variable", "field"])
    .dropna()
    .groupby(["date", "optimization"])
    .agg("mean")
    .reset_index()
    .rename(columns = {"optimization": ""})
    .pivot(index = "date", columns = "", values = "value")
    .apply(lambda x: np.cumprod(1 + x))
    .rename(columns = RENAMER)
    .plot(
        ax   = axes,
        logy = True,
        title = "Cumulative Returns Controlling for Sample Sizes (2018 Cutoff)\n10% Annualized Volatility Hedged Contemporaneously"))

plt.show()
```

```python
(df_group_port
    .drop(columns = ["field"])
    .groupby(["date", "optimization", "variable"])
    .agg("mean")
    .reset_index()
    .drop(columns = ["date"])
    .dropna()
    .groupby(["optimization", "variable"])
    .agg(lambda x: x.mean() / x.std() * np.sqrt(252))
    .reset_index()
    .replace(RENAMER)
    .rename(columns = {
        "optimization": "",
        "variable"    : "Vol"})
    .pivot(index = "", columns = "Vol", values = "value")
    .rename(columns = RENAMER))
```
