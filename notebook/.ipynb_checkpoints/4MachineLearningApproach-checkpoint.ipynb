# Machine Learning Factor Approach

```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
repo_path = os.path.abspath(os.path.join(os.getcwd(), ".."))
data_path = os.path.join(repo_path, "data")
back_path = os.path.join(data_path, "Backtests")
pic_path  = os.path.join(repo_path, "writeup", "pictures")
```

```python
model_renamer_ = {
    'elastic_net'  : "Elastic Net", 
    'polynomial'   : "Polynomial", 
    'random_forest': "Random Forest", 
    'ridge'        : "Ridge", 
    'xgboost'      : "XGBoost",
    "linear"       : "Linear"}

group_renamer_ = {
    'em_carry'        : "EM Carry",
    'g10_carry'       : "G10 Carry",
    'illiquid_future' : "Ill Fut",
    'liquid_future'   : "Liq Fut"}

vol_renamer_ = {
    "lagged"  : "Lagged",
    "perfect" : "Perfect",
    "cagr"    : "CAGR",	
    "mean_rtn": "Ann. Mean Rtn."}

sample_renamer_ = {
    "in_sample"  : "In-Sample",
    "out_sample" : "Out-Sample",
    "full_sample": "Full-Sample"}

renamer = {**model_renamer_, **group_renamer_, **sample_renamer_, **vol_renamer_}
FLOAT_FORMAT = lambda x: f"{x:.2f}"
```

```python
path      = os.path.join(data_path, "FXTickerGuide.xlsx")
df_ticker = (pd
    .read_excel(io = path, sheet_name = "TickerGuide")
    .rename(columns = {"ticker": "fx_ticker"})
    .loc[lambda x: x.Active == True]
    [["fx_ticker", "country", "group", "etf_ticker", "shorthand_name"]]
    .assign(etf_ticker = lambda x: x.etf_ticker.str.split(" ").str[0]))
```

```python
path = os.path.join(back_path, "MLFactorBacktest.parquet")

df_backtest = (pd
    .read_parquet(path = path, engine = "pyarrow")
    .drop(columns = ["name"])
    .drop(columns = ["fx_rtn", "signal_rtn"])
    .melt(id_vars = ["date", "forecasted", "fx_ticker", "model"])
    .dropna()
    .assign(target = lambda x: np.where(x.variable == "vol_rtn", "perfect", "lagged"))
    .drop(columns = ["variable"])
    .merge(right = df_ticker, how = "inner", on = ["fx_ticker"])
    .assign(
        year         = lambda x: x.date.dt.year,
        sample_group = lambda x: np.where(x.year <= 2018, "in_sample", "out_sample"))
    .drop(columns = ["year"]))
```

```python
display(df_backtest
    .loc[lambda x: x.target == "perfect"]
    .pivot(index = ["date", "fx_ticker", "sample_group"], columns = "model", values = "forecasted")
    .reset_index()
    .melt(id_vars = ["date", "fx_ticker", "sample_group", "linear"])
    .drop(columns = ["date"])
    .groupby(["fx_ticker", "sample_group", "model"])
    .agg("corr")
    .reset_index()
    .drop(columns = ["level_3", "linear"])
    .loc[lambda x: x.value != 1]
    .dropna()
    .merge(right = df_ticker, how = "inner", on = ["fx_ticker"])
    [["sample_group", "model", "value", "group"]]
    .groupby(["sample_group", "model", "group"])
    .agg("mean")
    .reset_index()
    .replace(renamer)
    .rename(columns = {
        "group"       : "Group",
        "sample_group": ""})
    .pivot(index = ["Group"], columns = ["", "model"], values = "value"))
```

```python
(df_backtest
    .loc[lambda x: x.target == "perfect"]
    [["date", "model", "value", "group", "sample_group"]]
    .groupby(["date", "model", "group", "sample_group"])
    .agg("mean")
    .reset_index()
    .drop(columns = ["date"])
    .groupby(["model", "group", "sample_group"])
    .agg(lambda x: x.mean() / x.std() * np.sqrt(252))
    .reset_index()
    .replace(renamer)
    .rename(columns = {
        "group"       : "Group",
        "model"       : "",
        "sample_group": "Sample Group"})
    .pivot(index = ["Group"], columns = ["", "Sample Group"], values = "value"))
```

```python
display(df_backtest
    .loc[lambda x: x.target == "perfect"]
    [["date", "model", "value", "group", "sample_group"]]
    .groupby(["date", "model", "group", "sample_group"])
    .agg("mean")
    .reset_index()
    .drop(columns = ["date"])
    .groupby(["model", "group", "sample_group"])
    .agg(lambda x: x.mean() / x.std() * np.sqrt(252))
    .reset_index()
    .replace(renamer)
    .rename(columns = {
        "group"       : "Group",
        "model"       : "",
        "sample_group": "Sample Group"})
    .pivot(index = [""], columns = ["Group", "Sample Group"], values = "value"))
```

```python
df_sharpe = (df_backtest
    [["date", "target", "model", "value", "sample_group"]]
    .groupby(["date", "target", "model", "sample_group"])
    .agg("mean")
    .reset_index()
    .drop(columns = ["date"])
    .groupby(["target", "model", "sample_group"])
    .agg(lambda x: x.mean() / x.std() * np.sqrt(252))
    .reset_index())
```

```python
display(df_sharpe
    .reset_index()
    .replace(renamer)
    .rename(columns = {
        "sample_group": "",
        "model"       : "Model",
        "target"      : "Vol. Target"})
    .pivot(index = "Model", columns = ["Vol. Target", ""], values = "value"))
```

```python
df_rtn = (df_backtest
    .loc[lambda x: x.sample_group == "out_sample"]
    .loc[lambda x: x.target == "perfect"]
    [["date", "model", "group", "value"]]
    .groupby(["date", "model", "group"])
    .agg("mean")
    .reset_index())

groups = df_rtn.group.drop_duplicates().sort_values().to_list()
fig, axes = plt.subplots(ncols = 2, nrows = 2, figsize = (16,8))

for group, ax in zip(groups, axes.flatten()):

    (df_rtn
        .loc[lambda x: x.group == group]
        .rename(columns = {"model": ""})
        .pivot(index = "date", columns = "", values = "value")
        .rename(columns = renamer)
        .apply(lambda x: np.cumprod(1 + x))
        .plot(
            ax     = ax,
            logy   = True,
            xlabel = "",
            ylabel = "Cuml. Ret. Log-Scaled",
            title  = renamer[group]))

fig.suptitle("Cumulative Returns of Out-of-Sample Returns by Model\n10% Annualized Volatility Targeting Hedged Contemporaneously\n")
#fig.savefig(os.path.join(pic_path, "OutSampleMLBacktest.jpeg"), dpi = 300)
plt.tight_layout()
```

```python
def _get_rtn_stat(df: pd.DataFrame) -> pd.DataFrame: 

    cagr = (df
        .set_index("date")
        [["value"]]
        .apply(lambda x: np.cumprod(1 + x) * 100)
        .reset_index()
        .loc[lambda x: (x.date == x.date.min()) | (x.date == x.date.max())]
        .assign(date_name = lambda x: np.where(x.date == x.date.min(), "min_date", "max_date"))
        .rename(columns = {
            "value": "px_val",
            "date" : "date_val"})
        .melt(id_vars = "date_name")
        .assign(name = lambda x: x.date_name + "_" + x.variable)
        .set_index("name")
        [["value"]]
        .T
        .assign(
            years = lambda x: (pd.to_datetime(x.max_date_date_val) - pd.to_datetime(x.min_date_date_val)).dt.days / 360,
            ratio = lambda x: x.max_date_px_val / x.min_date_px_val,
            cagr  = lambda x: x.ratio ** (1 / x.years) - 1)
        .cagr
        .item())

    avg_rtn = df.value.mean() * 252
    df_out  = (pd.DataFrame(
        data = {
            "cagr": [cagr],
            "mean_rtn": [avg_rtn]}))

    return df_out
```

```python
df_avg_rtn = (df_backtest
    .loc[lambda x: x.target == "perfect"]
    [["date", "model", "value", "group", "sample_group"]]
    .groupby(["date", "model", "group", "sample_group"])
    .agg("mean")
    .reset_index()
    .assign(sample_name = lambda x: x.model + "-" + x.group + "-" + x.sample_group)
    .groupby("sample_name")
    .apply(_get_rtn_stat)
    .reset_index()
    .drop(columns = ["level_1"])
    .assign(
        str_split    = lambda x: x.sample_name.str.split("-"),
        model        = lambda x: x.str_split.str[0],
        group        = lambda x: x.str_split.str[1],
        sample_group = lambda x: x.str_split.str[2])
    .drop(columns = ["sample_name", "str_split"]))
```

```python
(df_avg_rtn
    .melt(id_vars = ["model", "group", "sample_group"])
    .replace(renamer)
    .rename(columns = {
        "sample_group": "Sample Group",
        "variable"    : "",
        "model"       : "Model",
        "group"       : "Group"})
    .pivot(index = ["Model", "Group"], columns = ["", "Sample Group"], values = "value"))
```

```python
(df_sharpe
    .pivot(index = ["model", "target"], columns = "sample_group", values = "value")
    .assign(sharpe_spread = lambda x: x.out_sample - x.in_sample)
    .reset_index()
    .replace(renamer)
    .rename(columns = {"model": "Model", "target": ""})
    .pivot(index = "Model", columns = "", values = "sharpe_spread"))
```

```python
path = r"C:\Users\Diego\Documents\GitHub\FXEquitySpillover\data\MLModels\ModelMetrics.parquet"
df_rmse = (pd
    .read_parquet(path = path, engine = "pyarrow")
    .loc[lambda x: x.sample_group == "full_sample"]
    .pivot(index = "fx_ticker", columns = "model", values = "rmse")
    .reset_index()
    .melt(id_vars = ["fx_ticker", "linear"])
    .assign(
        rmse_spread = lambda x:  (x.value - x.linear)  / x.linear)
    [["fx_ticker", "rmse_spread", "model"]])
```

```python
path       = os.path.join(data_path, "MLModels", "ModelMetrics.parquet")
df_metrics = pd.read_parquet(path = path, engine = "pyarrow")

df_eval_change = (df_metrics
    .melt(id_vars = ["fx_ticker", "model", "sample_group"])
    .pivot(index = ["fx_ticker", "variable", "sample_group"], columns = "model", values = "value")
    .reset_index()
    .melt(id_vars = ["fx_ticker", "variable", "linear", "sample_group"])
    .assign(change = lambda x: x.value - x.linear))
```

```python
df_fullsample_sharpe = (df_backtest
    [["fx_ticker", "target", "value", "model"]]
    .groupby(["fx_ticker", "target", "model"])
    .agg(lambda x: x.mean() / x.std() * np.sqrt(252))
    .reset_index()
    .assign(sample_group = "full_sample"))

df_outsample_sharpe = (df_backtest
    .loc[lambda x: x.sample_group == "out_sample"]
    [["fx_ticker", "target", "value", "model"]]
    .groupby(["fx_ticker", "target", "model"])
    .agg(lambda x: x.mean() / x.std() * np.sqrt(252))
    .reset_index()
    .assign(sample_group = "out_sample"))

df_sharpe_compare = (pd
    .concat([df_fullsample_sharpe, df_outsample_sharpe])
    .rename(columns = {"value": "sharpe_spread"}))
```

```python
df_plot = (df_sharpe_compare
    .loc[lambda x: x.target == "perfect"]
    .merge(right = df_eval_change, how = "inner", on = ["fx_ticker", "sample_group", "model"])
    .loc[lambda x: x.variable == "rmse"]
    .loc[lambda x: ~x.model.isin(["ridge"])]
    .replace(renamer))

sample_groups = df_plot.sample_group.drop_duplicates().sort_values().to_list()
fig, axes     = plt.subplots(ncols = len(sample_groups), figsize = (20,6))

for sample_group, ax in zip(sample_groups, axes.flatten()): 

    df_tmp = (df_plot
        .loc[lambda x: x.sample_group == sample_group]
        .loc[lambda x: x.change < 0.001])

    sns.scatterplot(
        ax   = ax,
        data = df_tmp,
        x    = "change",
        y    = "sharpe_spread",
        hue  = "model")

    ax.set_title(sample_group)
    ax.set_xlabel("ML RMSE - Linear RMSE / ML RMSE")
    ax.set_ylabel("ML Sharpe - Linear Sharpe")

fig.suptitle("Comparing Relative RMSE Accuracy Relative to Relative Sharpe\nSome Outliers Moved (Excluding Ridge)")
plt.show()
```
